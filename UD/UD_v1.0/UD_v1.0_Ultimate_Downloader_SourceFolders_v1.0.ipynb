# Ultimate Downloader · Source Folders v1.0
*Based on Ultimate Downloader v7.3.*

This notebook embeds a local copy of `ultimate_downloader.py`. Nothing is fetched from GitHub at runtime.

With **Auto-organise** off, torrents are saved verbatim to `My Drive/0DL/0U_DL/<Torrent name>/<subfolders>/…`:
- Original folder layout and original file names.
- Nothing is dropped as a sample, image or tiny file.
- Archives (`.zip`/`.rar`/`.7z`) are saved as they are, never extracted.
- Subtitle names are not re-tagged.

**Queue preview:** use **List / Tree** above the queue. Tree view groups files by folder like a file manager: tap a folder to open or close it, tick its box to select everything inside, and use **Expand all / Collapse all**. The table is also taller now, so more rows are visible at once.

**Defaults in this build**
- Debrid: TorBox
- Upload to Drive via API: on
- Overlap Drive moves with downloads: on
- Auto-organise: off
- TMDB matching: off
- Downloads folder: `0DL/0U_DL`

**Tips**
- Store your TorBox key as a Colab Secret named `TB_TOKEN`, or paste it in **⚙️ Settings**.
- If you used an older version before, delete `My Drive/Ultimate Downloader/settings.json` once (or change the values in **⚙️ Settings**). Saved settings override the new defaults.
- Just run the cell below. The UI appears right away, and Drive mounts when you start.
